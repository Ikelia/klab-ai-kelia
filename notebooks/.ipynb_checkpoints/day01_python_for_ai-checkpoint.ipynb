```python
# Task 3.1: normalise() function
print("\n=== Normalise Function ===\n")

def normalise(values, minimum=None, maximum=None):
    """
    Scale a list of numbers into the 0-1 range.
    
    Args:
        values: List of numeric values to normalise
        minimum: Optional custom minimum value (defaults to min(values))
        maximum: Optional custom maximum value (defaults to max(values))
    
    Returns:
        List of normalised values in the range [0, 1]
    
    Example:
        normalise([12, 18, 25, 40]) → [0.0, 0.214, 0.464, 1.0]
    """
    lo = min(values) if minimum is None else minimum
    hi = max(values) if maximum is None else maximum
    span = (hi - lo) or 1  # Prevent division by zero if all values are equal
    
    return [(v - lo) / span for v in values]

# Test 1: Normal range of values
print("Test 1 - Normalise [12, 18, 25, 40]:")
result1 = normalise([12, 18, 25, 40])
print(f"Result: {[round(x, 3) for x in result1]}\n")

# Test 2: All identical values (should not crash)
print("Test 2 - Normalise identical values [5, 5, 5, 5]:")
result2 = normalise([5, 5, 5, 5])
print(f"Result: {result2} (all values are 0 because span is 1)\n")

# Test 3: With custom bounds
print("Test 3 - Normalise [12, 18, 25, 40] with custom min=10, max=50:")
result3 = normalise([12, 18, 25, 40], minimum=10, maximum=50)
print(f"Result: {[round(x, 3) for x in result3]}")
```

```python
# Task 3.2: summarise_scores() function
print("\n=== Summarise Scores Function ===\n")

def summarise_scores(scores):
    """
    Calculate summary statistics for a list of scores.
    
    Args:
        scores: List of numeric scores
    
    Returns:
        Dictionary containing:
            - count: number of scores
            - mean: average score
            - minimum: lowest score
            - maximum: highest score
            - above_threshold: count of scores >= 0.8
    
    If scores is empty, returns count=0, mean=None, min=None, max=None, above_threshold=0
    """
    if not scores:
        return {
            "count": 0,
            "mean": None,
            "minimum": None,
            "maximum": None,
            "above_threshold": 0
        }
    
    return {
        "count": len(scores),
        "mean": round(sum(scores) / len(scores), 4),
        "minimum": min(scores),
        "maximum": max(scores),
        "above_threshold": sum(1 for s in scores if s >= 0.8)
    }

# Test 1: Normal scores
print("Test 1 - Scores [0.71, 0.83, 0.88, 0.91, 0.79, 0.93]:")
print(summarise_scores([0.71, 0.83, 0.88, 0.91, 0.79, 0.93]))
print()

# Test 2: Empty list (should not crash)
print("Test 2 - Empty list []:")
print(summarise_scores([]))
print()

# Test 3: All scores above threshold
print("Test 3 - All high scores [0.85, 0.92, 0.88, 0.95]:")
print(summarise_scores([0.85, 0.92, 0.88, 0.95]))
```

```python
# Task 3.3: safe_divide() function
print("\n=== Safe Divide Function ===\n")

def safe_divide(numerator, denominator, default=0.0):
    """
    Safely divide two numbers with comprehensive error handling.
    
    Args:
        numerator: Number to be divided
        denominator: Number to divide by
        default: Value to return on error (defaults to 0.0)
    
    Returns:
        Division result as float, or default if an error occurs
    
    Handles:
        - Division by zero
        - Non-numeric inputs (strings, None, etc.)
        - TypeError from invalid operations
    
    Example:
        safe_divide(10, 2) → 5.0
        safe_divide(10, 0) → 0.0
        safe_divide("hello", 2) → 0.0
    """
    try:
        result = float(numerator) / float(denominator)
        return result
    except (TypeError, ValueError, ZeroDivisionError) as e:
        print(f" Error caught: {type(e).__name__}")
        return default


print("Test 1 - Normal division: safe_divide(10, 2)")
print(f"Result: {safe_divide(10, 2)}\n")


print("Test 2 - Division by zero: safe_divide(10, 0)")
print(f"Result: {safe_divide(10, 0)}\n")


print("Test 3 - Text numerator: safe_divide('hello', 2)")
print(f"Result: {safe_divide('hello', 2)}\n")


print("Test 4 - Text denominator: safe_divide(10, 'world')")
print(f"Result: {safe_divide(10, 'world')}\n")


print("Test 5 - Both non-numeric: safe_divide('abc', 'def')")
print(f"Result: {safe_divide('abc', 'def')}")
```

```python
# Task 4.2: Pandas
print("\n=== Pandas DataFrame ===\n")

import pandas as pd
import numpy as np


df = pd.DataFrame({
    "student": ["Aline", "Bosco", "Chantal", "David", "Eve", "Frank"],
    "experience": [1, 4, 3, None, 2, 5],  # Missing value
    "exam_score": [62, 91, 78, 55, 85, 70],
    "attendance": [90, 95, 85, 70, 92, 88],
    "completed": ["Yes", "Yes", "No", "No", "Yes", "Yes"]
})

print("=== DataFrame ===")
print(df)
print("\n" + "="*50 + "\n")


print("=== df.info() ===")
df.info()
print("\n" + "="*50 + "\n")

print("=== df.describe() ===")
print(df.describe())
print("\n" + "="*50 + "\n")

print("=== Missing Values (df.isna().sum()) ===")
print(df.isna().sum())
print("\n" + "="*50 + "\n")

print("=== Filter: Students with exam_score > 70 ===")
filtered = df[df["exam_score"] > 70].sort_values("exam_score", ascending=False)
print(filtered)
print("\n" + "="*50 + "\n")


df["passed"] = df["exam_score"] >= 70
print("=== DataFrame with 'passed' column ===")
print(df)
```

### Missing Values Explanation

The missing value in the "experience" column for student "David" could mean:
1. David is a new student with no prior experience (0 years)
2. The data was not recorded during collection
3. David chose not to disclose this information

In a real-world scenario, this missing value would need to be investigated. It could be filled with the mean experience (3 years) if the data is missing at random, or we might need to collect this information again from David.
